## 6. Encontrar os arquivos do curso

Descompacte o pacote inteiro. A pasta curso_pandas deve conter as pastas dados e notebooks lado a lado. A célula abaixo procura dados a partir da pasta atual ou da pasta acima. Path é uma ferramenta do Python para trabalhar com caminhos. Aqui ela prepara a localização dos arquivos; você não precisa estudar todos os seus detalhes para começar pandas.

O resultado deve terminar em dados/penguins.csv e indicar True. Se indicar False ou interromper com uma mensagem, confira se você abriu o notebook dentro do pacote descompactado e se a pasta dados está presente.


```python
from pathlib import Path
BASE = Path.cwd()
if not (BASE / 'dados').exists() and (BASE.parent / 'dados').exists():
    BASE = BASE.parent
DADOS = BASE / 'dados'
arquivo = DADOS / 'penguins.csv'
print(arquivo)
print('Arquivo encontrado:', arquivo.exists())
assert arquivo.exists(), 'Confira se dados e notebooks estão dentro da pasta curso_pandas.'
```

## 7. Ler e observar

Agora vamos ler o arquivo. A primeira linha guarda a tabela no nome p. A segunda mostra três registros. Não tente entender todas as colunas neste momento. Confira apenas: existem nomes de colunas no topo? Existem três linhas? Você consegue localizar species e body_mass_g?


```python
p = pd.read_csv(arquivo)
print(p.head(3))
```
